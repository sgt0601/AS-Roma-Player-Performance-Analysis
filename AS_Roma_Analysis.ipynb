# Part5: DATA SCIENCE

The question I want to further explore is "Which players overperform or underperform relative to their expected goals?". From the observations above, we can see that players with higher expected goals tend to have higher performance, but there are also players who overperform or underperform. Since this question involves predicting one variable from another, this is really appropriate to use a machine learning model, especially a linear regression model.

The reason why I choose this question is that the expected goals is one of the most used measures in modern football, and it provides people a way to measure the quality of scoring chances, and from the analysis, we can see which player can better hold the chance to score a goal, and which players are inefficient goal scorer.

```python
from sklearn.linear_model import LinearRegression
df = df.fillna(0)
df = df[df.MP >= 10]
df["Performance"] = df["Gls"] + df["Ast"]
ind = df[["xG"]]
dep = df["Performance"]
model = LinearRegression()
model.fit(ind, dep)
df["PredictedPerformance"] = model.predict(ind)
df
```

Recall that I have defined the performance as the sum of goals and assists, and I used a linear regression model using expected goals as the independent variable and the players performance as the dependent variable, then the new column is the predicted performance values for each players. After fitting the model, I want to then calculate the residual:

```python
df["Residual"] = df["Performance"] - df["PredictedPerformance"]
df
```

From the residual, we can see there is negative values and positive values, if it is negative values, then it represents that the player's performance is worse than the predicted performance. However, when it is positive, it represents that the player's performance is better than the predicted performance.

```python
overperformance = df[df["Residual"] > 0]
overperformance
```

```python
underperformance = df[df["Residual"] < 0]
underperformance
```

So from this, we have found which players performed above the expectations and which players performed below the expectations.
